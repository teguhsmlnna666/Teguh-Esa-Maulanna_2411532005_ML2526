```python
import numpy as np
import pandas as pd

from sklearn.model_selection import KFold, cross_val_score
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

```python
from google.colab import drive
drive.mount('/content/drive')

submission_path = "/content/drive/MyDrive/ML/Dataset/gender_submission.csv"
test_path = "/content/drive/MyDrive/ML/Dataset/test.csv"
train_path = "/content/drive/MyDrive/ML/Dataset/train.csv"

train_df = pd.read_csv(train_path)

X = train_df.drop('Survived', axis=1)
y = train_df['Survived']
```

```python
# hapus kolom yang tidak digunakan
train_df = train_df.drop(['PassengerId', 'Name', 'Ticket', 'Cabin'], axis=1)

# Atasi missing value
train_df['Age'] = train_df['Age'].fillna(train_df['Age'].mean())
train_df['Embarked'] = train_df['Embarked'].fillna(train_df['Embarked'].mode()[0])

# Encoding data
le = LabelEncoder()
train_df['Sex'] = le.fit_transform(train_df['Sex'])
train_df['Embarked'] = le.fit_transform(train_df['Embarked'])
```

```python
model = LogisticRegression(max_iter=1000)

X = train_df.drop('Survived', axis=1)
y = train_df['Survived']

for k in [5, 10]:
    print(f"K-Fold (k={k})")

    accuracy = cross_val_score(model, X, y, cv=k, scoring='accuracy')
    precision = cross_val_score(model, X, y, cv=k, scoring='precision')
    recall = cross_val_score(model, X, y, cv=k, scoring='recall')
    f1 = cross_val_score(model, X, y, cv=k, scoring='f1')

    print("Accuracy tiap fold:", accuracy)
    print("Accuracy rata-rata:", accuracy.mean())

    print("Precision rata-rata:", precision.mean())
    print("Recall rata-rata:", recall.mean())
    print("F1-score rata-rata:", f1.mean())
```
